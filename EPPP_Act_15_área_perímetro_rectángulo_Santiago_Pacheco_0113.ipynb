## Redaccion del problema
* ✍ pasos en la resolución de problemas geométricos mediante código. Un rectángulo es una figura plana de dos dimensiones con cuatro lados y cuatro ángulos rectos. Al ser una figura 2D, solo podemos calcular su área (espacio que ocupa en el plano) y su perímetro (longitud total de sus bordes).El programa utiliza una estructura de menú interactivo que se mantiene activo hasta que el usuario decida salir. Se apoya en un ciclo while para repetir la ejecución, condicionales if/elif/else para dirigir el flujo según la opción elegida, y llamadas a funciones que encapsulan cada cálculo. Esto hace que el código sea modular, fácil de leer y mantener. En cada función se solicitan las medidas necesarias (base y altura), se aplica la fórmula correspondiente y se muestra el resultado de forma clara y elegante.
## Formulas a utilizar
1. Area de un triangulo
* $A=b×h$
2. Perimetro de un rectangulo
* $P=2×(b+h)$

## Caracteristicas del problema❗
* Codigo en un solo archivolisto para ejecutar
* Sin programacion orientada a objetos
* Resultados formateados con 2 decimales
* solicitud de base y altura dentro de cada funcion
* Llamadas a funciones desde el menu principal
* Uso de estructura de control: if, elif, else, while
* Menu interactivo con emojis y diseño elegante
* Estructura modular con 3 funciones independientes

```python
# 📐 Programa interactivo para calcular área y perímetro de un rectángulo
# Nivel: Principiante (Preparatoria)
# Estructura: Procedural (sin POO), sin validación de entrada, con menú cíclico

def area_rectangulo():
    """Calcula el área de un rectángulo pidiendo base y altura."""
    print("\n📏 Cálculo del Área del Rectángulo")
    base = float(input("   Ingresa la base del rectángulo: "))
    altura = float(input("   Ingresa la altura del rectángulo: "))

    area = base * altura
    print(f"✅ El área del rectángulo es: {area} unidades cuadradas\n")

def perimetro_rectangulo():
    """Calcula el perímetro de un rectángulo pidiendo base y altura."""
    print("\n🔲 Cálculo del Perímetro del Rectángulo")
    base = float(input("   Ingresa la base del rectángulo: "))
    altura = float(input("   Ingresa la altura del rectángulo: "))

    perimetro = 2 * (base + altura)
    print(f"✅ El perímetro del rectángulo es: {perimetro} unidades lineales\n")

def menu():
    """Muestra un menú interactivo y repite hasta que el usuario elija salir."""
    while True:
        print("\n" + "="*40)
        print("📐 MENÚ DE CÁLCULO GEOMÉTRICO 📐")
        print("="*40)
        print("1️⃣  Calcular área del rectángulo")
        print("2️⃣  Calcular perímetro del rectángulo")
        print("3️⃣  Salir del programa")
        print("="*40)

        opcion = input("👉 Selecciona una opción (1-3): ")

        if opcion == "1":
            area_rectangulo()          # Llamada a función
        elif opcion == "2":
            perimetro_rectangulo()     # Llamada a función
        elif opcion == "3":
            print("\n👋 ¡Gracias por usar el programa! Hasta pronto 🌟")
            print("programa realizado por Santiago Pacheco NL = 43")
            break                      # Sale del ciclo while
        else:
            print("\n⚠️ Opción no válida. Por favor, elige 1, 2 o 3.\n")

# Punto de entrada principal
if __name__ == "__main__":
    menu()
```
